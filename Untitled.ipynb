```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson, norm
df = pd.read_excel("Online Retail.xlsx")
display(df.head())
df.info()

display(df.describe())
```

```python
print("Dataset shape:", df.shape)
df.info()
display(df.describe())
```

```python
df = df.dropna(subset=["Description", "CustomerID"])
df = df.drop_duplicates()
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)].copy()
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["Month"] = df["InvoiceDate"].dt.month

print("Cleaned dataset shape:", df.shape)

display(df[["Quantity", "UnitPrice", "TotalPrice"]].describe())
```

```python
country_sales = (
    df.groupby("Country")["TotalPrice"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 5))
country_sales.plot(kind="bar")
plt.title("Top 10 Countries by Total Sales")
plt.xlabel("Country")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
plt.figure(figsize=(8, 5))
plt.hist(df["Quantity"], bins=50)
plt.title("Distribution of Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.xlim(0, 100)
plt.tight_layout()
plt.show()
monthly_sales = df.groupby(
    df["InvoiceDate"].dt.to_period("M")
)["TotalPrice"].sum()

plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index.astype(str),
         monthly_sales.values, marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import binom
p = (df["Country"] == "United Kingdom").mean()
n = 20
k = np.arange(0, n + 1)

probabilities = binom.pmf(k, n, p)

plt.figure(figsize=(9, 5))
plt.bar(k, probabilities)
plt.title("Binomial Distribution - UK Sales")
plt.xlabel("Number of UK Sales in 20 Trials")
plt.ylabel("Probability")
plt.tight_layout()
plt.show()

print("Probability of UK sale:", round(p, 4))
print("Expected number:", round(n * p, 2))
```

```python
# Poisson Distribution
from scipy.stats import poisson

# Count unique invoices per day
daily_orders = df.groupby(
    df["InvoiceDate"].dt.date
)["InvoiceNo"].nunique()

lam = daily_orders.mean()

k = np.arange(0, int(lam + 4 * np.sqrt(lam)) + 1)

poisson_probs = poisson.pmf(k, lam)

plt.figure(figsize=(9, 5))
plt.bar(k, poisson_probs)
plt.title("Poisson Distribution - Daily Orders")
plt.xlabel("Number of Orders Per Day")
plt.ylabel("Probability")
plt.tight_layout()
plt.show()

print("Average daily orders:", round(lam, 2))
print("Probability of 100 orders:",
      poisson.pmf(100, lam))
```

```python

# NORMAL DISTRIBUTION

from scipy.stats import norm

prices = df["UnitPrice"]

mu = prices.mean()
sigma = prices.std()

x = np.linspace(
    prices.quantile(0.01),
    prices.quantile(0.99),
    300
)

plt.figure(figsize=(10, 5))

plt.hist(
    prices,
    bins=60,
    density=True,
    alpha=0.5,
    label="Actual Data"
)

plt.plot(
    x,
    norm.pdf(x, mu, sigma),
    color="red",
    linewidth=2,
    label="Normal Distribution"
)

plt.xlim(
    prices.quantile(0.01),
    prices.quantile(0.99)
)

plt.title("Normal Distribution of Unit Prices")
plt.xlabel("Unit Price")
plt.ylabel("Density")
plt.legend()
plt.tight_layout()
plt.show()

print("Mean:", round(mu, 2))
print("Standard Deviation:", round(sigma, 2))
print("Skewness:", round(prices.skew(), 2))
```

```python

# COMPARISON OF DISTRIBUTIONS

print("=== BINOMIAL DISTRIBUTION ===")
print("Mean:", round(n * p, 2))
print("Variance:", round(n * p * (1 - p), 2))

print("\n=== POISSON DISTRIBUTION ===")
print("Mean:", round(lam, 2))
print("Observed Variance:", round(daily_orders.var(), 2))
print("Variance-to-Mean Ratio:",
      round(daily_orders.var() / lam, 2))

print("\n=== NORMAL DISTRIBUTION ===")
print("Mean:", round(mu, 2))
print("Standard Deviation:", round(sigma, 2))
print("Skewness:", round(prices.skew(), 2))
```

```python

# EFFECT OF LARGER SAMPLE SIZES

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Binomial: increasing number of trials
for trials in [20, 100, 500]:
    values = np.arange(trials + 1)
    probs = binom.pmf(values, trials, p)

    axes[0].plot(
        values / trials,
        probs * trials,
        label=f"n={trials}"
    )

axes[0].set_title("Binomial: Increasing Sample Size")
axes[0].set_xlabel("Proportion of UK Sales")
axes[0].set_ylabel("Scaled Probability")
axes[0].legend()

# Poisson: increasing expected event count
for rate in [5, 20, 60]:
    values = np.arange(0, int(rate + 5*np.sqrt(rate)) + 1)
    probs = poisson.pmf(values, rate)

    axes[1].plot(
        (values-rate)/np.sqrt(rate),
        probs*np.sqrt(rate),
        label=f"Lambda={rate}"
    )

axes[1].set_title("Poisson: Increasing Rate")
axes[1].set_xlabel("Standardised Order Count")
axes[1].set_ylabel("Scaled Probability")
axes[1].legend()

plt.tight_layout()
plt.show()
```
