```python
import numpy as np
from scipy import stats

# Final grades
grades = df["G3"]

# Sample statistics
mean_grade = grades.mean()
std_error = stats.sem(grades)
n = len(grades)

print("Sample Size:", n)
print("Mean Final Grade:", round(mean_grade, 2))

# Confidence intervals using t-distribution
for confidence in [0.90, 0.95, 0.99]:
    lower, upper = stats.t.interval(
        confidence,
        df=n-1,
        loc=mean_grade,
        scale=std_error
    )

    print(
        f"{confidence*100:.0f}% Confidence Interval:",
        f"({lower:.2f}, {upper:.2f})",
        f"Width: {upper-lower:.2f}"
    )
```

### Interpretation of Confidence Intervals

The dataset contains 649 students, with a sample mean final grade (G3) of approximately 11.91.

Confidence intervals were calculated at three confidence levels using the t-distribution.

- **90% Confidence Interval:** 11.70–12.11 (width = 0.42)
- **95% Confidence Interval:** 11.66–12.16 (width = 0.50)
- **99% Confidence Interval:** 11.58–12.23 (width = 0.66)

The results demonstrate that increasing the confidence level leads to a wider confidence interval. This occurs because a higher confidence level requires a larger critical value, increasing the margin of error.

The 90% confidence interval is the narrowest, while the 99% confidence interval is the widest.

The 95% confidence interval provides a commonly used balance between confidence and precision.

These intervals estimate the population mean final grade, assuming the sample is representative of the population of interest. The large sample size supports the use of a t-based confidence interval, although the sampling method and unusual observations should also be considered.

```python

import matplotlib.pyplot as plt
from scipy import stats

# Calculate confidence intervals
confidence_levels = [0.90, 0.95, 0.99]
intervals = []

for level in confidence_levels:
    lower, upper = stats.t.interval(
        level,
        df=n-1,
        loc=mean_grade,
        scale=std_error
    )
    intervals.append((lower, upper))

# Plot confidence intervals
plt.figure(figsize=(9, 5))

for i, (lower, upper) in enumerate(intervals):
    plt.plot(
        [lower, upper],
        [i, i],
        marker="|",
        markersize=12,
        linewidth=3
    )

plt.axvline(
    mean_grade,
    color="red",
    linestyle="--",
    label="Sample Mean"
)

plt.yticks(
    range(3),
    ["90%", "95%", "99%"]
)

plt.xlabel("Mean Final Grade (G3)")
plt.ylabel("Confidence Level")
plt.title("Comparison of Confidence Intervals")
plt.legend()
plt.grid(axis="x", alpha=0.3)

plt.show()
```

### Interpretation of Confidence Interval Visualisation

The graph compares the 90%, 95%, and 99% confidence intervals for the population mean final grade (G3).

All three confidence intervals are centred around the sample mean of approximately 11.91.

The 90% confidence interval is the narrowest, while the 99% confidence interval is the widest. This demonstrates the trade-off between confidence and precision.

As the confidence level increases, the critical value increases, resulting in a larger margin of error and a wider confidence interval.

These findings demonstrate how the choice of confidence level influences the precision of statistical estimates.

```python

from scipy import stats

# Separate final grades by gender
female = df[df["sex"] == "F"]["G3"]
male = df[df["sex"] == "M"]["G3"]

# Descriptive statistics for each group
print("Female students:", len(female))
print("Male students:", len(male))

print("Female mean:", round(female.mean(), 2))
print("Male mean:", round(male.mean(), 2))

# Check variance equality using Levene's test
levene_stat, levene_p = stats.levene(female, male)

print("Levene's test p-value:", round(levene_p, 4))

# Welch's independent samples t-test
t_stat, p_value = stats.ttest_ind(
    female,
    male,
    equal_var=False
)

print("t-statistic:", round(t_stat, 4))
print("p-value:", round(p_value, 4))
```

### Test 1: Independent Samples t-test

**Research Question:**

Is there a statistically significant difference in final grades between female and male students?

**Hypotheses:**

- **H₀ (Null Hypothesis):** There is no difference in the population mean final grades between female and male students.
- **H₁ (Alternative Hypothesis):** There is a difference in the population mean final grades between female and male students.

**Significance Level:** α = 0.05

**Test Selection:**

Welch's independent samples t-test was used to compare the mean final grades of two independent groups. This test does not require equal population variances.

**Results:**

- Female students: n = 383, mean = 12.25
- Male students: n = 266, mean = 11.41
- Levene's test p-value = 0.9327
- t-statistic = 3.2747
- p-value = 0.0011

**Interpretation:**

The p-value (0.0011) is lower than the significance level of 0.05. Therefore, the null hypothesis is rejected.

Female students achieved a higher average final grade than male students, with a difference of approximately 0.84 points.

Levene's test did not indicate a significant difference in variances between the two groups (p = 0.9327).

**Conclusion:**

The results suggest a statistically significant difference in mean final grades between female and male students in this dataset. However, this finding does not establish a causal relationship. The assumptions of independence and normality should also be evaluated.
